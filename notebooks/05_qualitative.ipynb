# 05 — Qualitative Retrieval


Hiển thị query và Top-K ảnh retrieved để trả lời câu hỏi: kết quả đúng có xuất hiện ở top đầu không, và lỗi thường do sản phẩm tương tự hình thức hay do background/quality?

```python
from pathlib import Path
import numpy as np, pandas as pd
from PIL import Image
import matplotlib.pyplot as plt

def show_query_results(method="proposed", query_row=0, k=10):
    df=pd.read_csv("data/sampled/sop_20k.csv")
    q=df[df.split=="query"].reset_index(drop=True)
    g=df[df.split=="gallery"].reset_index(drop=True)
    z=np.load(f"outputs/{method}/retrieval_results.npz")
    ids=z["ids"][query_row,:k]
    qpath=Path("data/raw/Stanford_Online_Products")/q.loc[query_row,"path"]
    fig,ax=plt.subplots(1,k+1,figsize=(3*(k+1),3))
    ax[0].imshow(Image.open(qpath).convert("RGB")); ax[0].set_title(f"QUERY\nclass={q.loc[query_row,'class_id']}"); ax[0].axis("off")
    for j,idx in enumerate(ids,1):
        p=Path("data/raw/Stanford_Online_Products")/g.loc[int(idx),"path"]
        ax[j].imshow(Image.open(p).convert("RGB"))
        ax[j].set_title(f"rank {j}\nclass={g.loc[int(idx),'class_id']}")
        ax[j].axis("off")
    plt.tight_layout(); plt.show()

show_query_results("proposed", 0, 5)
```
