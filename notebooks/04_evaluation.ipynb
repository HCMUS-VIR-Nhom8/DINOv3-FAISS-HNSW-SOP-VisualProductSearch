# 04 — Evaluation


Tập trung vào 4 nhóm chỉ số: Recall@K, mAP, latency/query và memory.

```python
import json, pandas as pd
rows=[]
for method in ["baseline","proposed"]:
    m=json.load(open(f"outputs/{method}/metrics.json"))
    m["method"]=method
    rows.append(m)
cmp=pd.DataFrame(rows).set_index("method")
display(cmp)
```

```python
# Xuất bảng thesis
cmp.to_csv("outputs/comparison/metrics_comparison.csv")
display(cmp.T)
```
