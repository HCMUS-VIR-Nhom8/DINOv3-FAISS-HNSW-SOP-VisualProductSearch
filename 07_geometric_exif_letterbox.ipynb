# 07. Geometric Normalization

Operations:

1. EXIF orientation correction
2. Orientation normalization
3. Aspect-ratio-preserving resize
4. Letterbox padding

Mục tiêu:
- tránh ảnh bị xoay sai
- tránh geometric distortion
- giữ nguyên hình dạng sản phẩm
- đưa input về kích thước cố định cho DINOv3

```python
from PIL import Image, ImageOps

def load_exif_corrected(path):
    image = Image.open(path)
    image = ImageOps.exif_transpose(image)
    return image.convert("RGB")
```

```python
def letterbox_pil(image, size=224, fill=0):
    w, h = image.size
    scale = min(
        size / w,
        size / h
    )

    new_w = round(w * scale)
    new_h = round(h * scale)

    image = image.resize(
        (new_w, new_h),
        Image.Resampling.LANCZOS
    )

    canvas = Image.new(
        "RGB",
        (size, size),
        color=(fill, fill, fill)
    )

    x = (size - new_w) // 2
    y = (size - new_h) // 2

    canvas.paste(
        image,
        (x, y)
    )

    return canvas
```
