# 06. Object Crop + Smart Padding

Input:
- original image
- object bounding box

Output:
- cropped product
- padded crop
- object coverage ratio
- crop coordinates

Padding được tính theo kích thước bounding box thay vì số pixel cố định.

```python
def smart_crop(image,bbox,padding_ratio=0.10):

    h, w = image.shape[:2]
    x1, y1, x2, y2 = bbox

    bw = x2 - x1
    bh = y2 - y1

    px = int(bw * padding_ratio)

    py = int(bh * padding_ratio)

    x1 = max(0, x1 - px)
    y1 = max(0, y1 - py)

    x2 = min(w, x2 + px)
    y2 = min(h, y2 + py)

    crop = image[
        y1:y2,
        x1:x2
    ]

    return crop, (x1,y1,x2,y2)
```

## Object coverage

```python
def bbox_area_ratio(bbox,image_width,image_height):

    x1, y1, x2, y2 = bbox
    bbox_area = (
        max(0, x2 - x1)
        *
        max(0, y2 - y1)
    )
    image_area = (
        image_width
        *
        image_height
    )

    return bbox_area / image_area
```
